```python
import h5py
import os
import glob

# Get only the _ik.h5 files in the directory
h5_files = glob.glob("/root/vast/eric/stac-mjx/reaches_*_ik.h5")

# List to store files with length > 150
long_clips = []

# Process each file
print("File path | qpos shape")
print("-" * 50)

for file_path in h5_files:
    try:
        with h5py.File(file_path, 'r') as f:
            if 'qpos' in f:
                shape = f['qpos'].shape
                print(f"{file_path} | {shape}")
                
                # Check if length is > 150
                if shape[0] > 150:
                    long_clips.append(file_path)
            else:
                print(f"{file_path} | No 'qpos' dataset found")
    except Exception as e:
        print(f"{file_path} | Error: {e}")

print("\n\nFiles with length > 150:")
print("-" * 50)
for clip in long_clips:
    print(clip)

print(f"\nTotal long clips found: {len(long_clips)}")

# Create the comma-separated string of long clips (no line breaks)
long_clips_string = ",".join(long_clips)
print("\nComma-separated list format:")
print(f'"{long_clips_string}"')
```
